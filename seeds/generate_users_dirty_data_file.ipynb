```python
import pandas as pd
import numpy as np
from faker import Faker
import random
```

```python
fake = Faker()
```

```python
tr_04_08 = pd.read_csv('../seeds/input_data/transaction_04_08_2025.csv')
tr_11_08 = pd.read_csv('../seeds/input_data/transaction_11_08_2025.csv')
tr_17_08 = pd.read_csv('../seeds/input_data/transaction_17_08_2025.csv')
tr_25_08 = pd.read_csv('../seeds/input_data/transaction_25_08_2025.csv')

country_codes = ('UK', 'AU', 'US', 'CA', 'DE')
```

```python
all_files_df = pd.concat([
    tr_04_08, tr_11_08,
    tr_17_08, tr_25_08
])

## Replace it after finishing transformation

# all_files_df = pd.concat([
#     tr_04_08['user_id'], tr_11_08['user_id'],
#     tr_17_08['user_id'], tr_25_08['user_id']
# ])
```

```python
all_files_df
```

```python
res_df = (all_files_df[['user_id']]
          .sort_values(by = 'user_id', ascending = True))
```

```python
res_df['first_name'] = [fake.first_name() for _ in range(len(res_df))]
```

```python
res_df[['first_name', 'last_name']] = [
    (fake.first_name(), fake.last_name())
    for _ in range(len(res_df))
]
```

```python
res_df.head()
```

```python
res_df['email'] = res_df['first_name'].str.lower() + '.'\
                   + res_df['last_name'].str.lower() + '@example.com'
```

```python
res_df['country'] = [random.choice(country_codes) for _ in range(len(res_df))]
```
