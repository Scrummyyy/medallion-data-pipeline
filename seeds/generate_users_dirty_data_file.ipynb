```python
import pandas as pd
import numpy as np
from faker import Faker
#import datetime
import random
```

```python
## Creating class

fake = Faker()
```

```python
## Reading files

tr_04_08 = pd.read_csv('../seeds/input_data/transaction_04_08_2025.csv')
tr_11_08 = pd.read_csv('../seeds/input_data/transaction_11_08_2025.csv')
tr_17_08 = pd.read_csv('../seeds/input_data/transaction_17_08_2025.csv')
tr_25_08 = pd.read_csv('../seeds/input_data/transaction_25_08_2025.csv')

## Declaring variables

country_codes = ('UK', 'AU', 'US', 'CA', 'DE')
referral_source_values = ("google_ads", "organic",  "social_media",
     "referral", "partner" )
```

```python
#tr_04_08.columns
```

```python
all_files_df = pd.concat([
    tr_04_08, tr_11_08,
    tr_17_08, tr_25_08
])

## Replace it after finishing transformation

# all_files_df = pd.concat([
#     tr_04_08[['user_id']], tr_11_08[[['user_id']],
#     tr_17_08[['user_id']], tr_25_08[['user_id']]
# ])
```

```python
## Df for transformation
res_df = all_files_df[['user_id']]
len_res_df = len(res_df)
```

```python
#res_df['first_name'] = [fake.first_name() for _ in range(len_res_df)]
```

```python
## Generating names with faker
res_df[['first_name', 'last_name']] = [
    (fake.first_name(), fake.last_name())
    for _ in range(len_res_df)
]
```

```python
res_df.head()
```

```python
## Generating emails according to name

res_df['email'] = res_df['first_name'].str.lower() + '.'\
                   + res_df['last_name'].str.lower() + '@example.com'
```

```python
## Generating dates with pandas

dates = pd.date_range("2023/01/01", "2024/01/01")
res_df["signup_date"] = pd.Series(dates).sample(n=len_res_df, replace=True).values
res_df["signup_date"] = [random.choice(dates) for _ in range(len_res_df)]
```

```python
## Generating country and source with random

res_df['country'] = [random.choice(country_codes) for _ in range(len_res_df)]
```

```python
## Generating source with numpy (the cheapest way)

res_df['referral_source'] = np.random.choice(referral_source_values, size = len_res_df)
```
