```python
import pandas as pd
import numpy as np
from faker import Faker
```

```python
## Creating class
fake = Faker()
```

```python
## Reading files

tr_04_08 = pd.read_csv('../data/transactions_01_06_2025.csv')
tr_11_08 = pd.read_csv('../data/transactions_01_13_2025.csv')
tr_17_08 = pd.read_csv('../data/transactions_01_20_2025.csv')
tr_25_08 = pd.read_csv('../data/transactions_01_27_2025.csv')

## Declaring variables

country_codes = ('UK', 'AU', 'US', 'CA', 'DE')
referral_source_values = (
    "google_ads", "organic",  "social_media",
     "referral", "partner"
)
```

```python
tr_04_08.columns
```

```python
## Replace it after finishing transformation
## Df for transformation

res_df = pd.concat([
    tr_04_08[['transaction_id']], tr_11_08[['transaction_id']],
    tr_17_08[['transaction_id']], tr_25_08[['transaction_id']]
], ignore_index = True) ## Avoiding ValueError: cannot reindex on an axis with duplicate labels

## Length variable
len_res_df = res_df.shape[0]
```

```python
## Memory cleaning
del tr_04_08, tr_11_08, tr_17_08, tr_25_08
```

```python
## Generating names with faker
res_df[['first_name', 'last_name']] = [
    (fake.first_name(), fake.last_name())
    for _ in range(len_res_df)
]
```

```python
res_df.head()
```

```python
## Generating emails according to name

res_df['email'] = res_df['first_name'].str.lower() + '.'\
                + res_df['last_name'].str.lower() + '@example.com'
```

```python
## Generating dates with pandas

dates = pd.date_range("2023/06/01", "2024/06/01")
range_1000_2000 = [i for i in range(1000, 2000)]

res_df["signup_date"] = pd.Series(dates).sample(n=len_res_df, replace=True).values
del dates
```

```python
## Generating user_id

res_df ["user_id"] = np.random.choice(range_1000_2000, size=len_res_df)

del range_1000_2000

res_df["user_id"] = "USR_" + res_df["user_id"].astype("str")
```

```python
## Generating country and source with numpy

res_df['country'] = np.random.choice(country_codes, size = len_res_df)
```

```python
## Generating source with numpy (the cheapest way)

res_df['referral_source'] = np.random.choice(referral_source_values, size = len_res_df)
```

```python
## Saving new df

res_df.to_csv(path_or_buf = '../data/users_data.csv', index = False)
```
